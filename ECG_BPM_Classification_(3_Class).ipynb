```
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout, BatchNormalization
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import wfdb  # Library for reading PhysioNet databases
from scipy.signal import find_peaks
import os

# --- 1. Project Parameters ---
SAMPLE_RATE = 360  # MIT-BIH database sampling rate is 360 Hz
WINDOW_SECONDS = 10  # 10-second ECG strips
WINDOW_SAMPLES = SAMPLE_RATE * WINDOW_SECONDS  # 3600 samples
CLASSES = ["Low (<60)", "Normal (60-100)", "High (>100)"]
N_CLASSES = len(CLASSES)

# --- 2. Data Download and Preparation ---

def download_mitdb():
    """Downloads the MIT-BIH Arrhythmia Database if not already present."""
    db_dir = 'mit-bih-arrhythmia-database-1.0.0'
    if not os.path.exists(db_dir):
        print("Downloading MIT-BIH Arrhythmia Database...")
        wfdb.dl_database('mitdb', '.')
        print("Download complete.")
    else:
        print("Database already downloaded.")
    return db_dir

def get_bpm_label(r_peaks, sampling_rate):
    """
    Rule-based method to calculate mean BPM from R-peaks and assign a label.
    """
    if len(r_peaks) < 2:
        return None  # Not enough peaks to calculate BPM

    # Calculate RR intervals in samples
    rr_intervals = np.diff(r_peaks)

    # Convert RR intervals to seconds
    rr_seconds = rr_intervals / sampling_rate

    # Calculate BPM for each interval
    bpms = 60 / rr_seconds

    # Calculate the mean BPM for the window
    mean_bpm = np.mean(bpms)

    # Assign label based on BPM
    if mean_bpm < 60:
        return 0  # Abnormally Low
    elif 60 <= mean_bpm <= 100:
        return 1  # Normal
    else:
        return 2  # Abnormally High

def process_data(db_dir):
    """
    Loads MIT-BIH data, segments it, and auto-labels it based on BPM.
    """
    # Get all record names in the database
    records = wfdb.get_record_list('mitdb')

    X = []  # To store 10-second signal windows
    y = []  # To store the corresponding labels (0, 1, or 2)

    print(f"Processing {len(records)} records...")

    for rec_name in records:
        # Load the record and use the MLII signal (usually channel 0)
        try:
            # Pass the directory to rdrecord
            record = wfdb.rdrecord(rec_name, sampfrom=0, sampto=5000, pn_dir='mitdb') # Changed to use pn_dir for PhysioNet database
            # Use channel 0 (MLII) as it's typically cleaner
            signal = record.p_signal[:, 0]
        except Exception as e:
            print(f"Could not read record {rec_name}. Skipping. Error: {e}")
            continue

        # Slide a 10-second window across the signal
        # We can use a stride of 5 seconds to create overlapping samples
        stride_samples = SAMPLE_RATE * 5  # 5-second stride

        for i in range(0, len(signal) - WINDOW_SAMPLES, stride_samples):
            window = signal[i : i + WINDOW_SAMPLES]

            # Normalize the window (important for peak detection and model)
            if np.max(window) - np.min(window) > 1e-6: # Avoid division by zero
                window_norm = (window - np.min(window)) / (np.max(window) - np.min(window))
            else:
                continue # Skip flat-line windows

            # --- Rule-Based Labeling ---
            # Find R-peaks. We set a min height and distance to be robust.
            # Min distance: 0.4s * 360Hz = 144 samples (max 150 BPM)
            r_peaks, _ = find_peaks(window_norm, height=0.5, distance=144)

            # Get the label (0, 1, or 2)
            label = get_bpm_label(r_peaks, SAMPLE_RATE)

            if label is not None:
                X.append(window_norm)
                y.append(label)

    print(f"Processing complete. Generated {len(X)} samples.")

    # Convert to numpy arrays
    X = np.array(X)
    y = np.array(y)

    # Reshape X for the 1D CNN: (num_samples, num_timesteps, num_features)
    X = X.reshape(X.shape[0], X.shape[1], 1)

    return X, y

# --- 3. Load and Prepare Dataset ---
db_dir = download_mitdb()
X, y = process_data(db_dir)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print(f"Training samples: {X_train.shape[0]}")
print(f"Test samples: {X_test.shape[0]}")
print(f"Input shape: {X_train.shape[1:]}")

# --- 4. Build Lightweight 1D-CNN Model ---
model = Sequential()

# Input Layer
model.add(Conv1D(filters=32, kernel_size=5, activation='relu',
                 input_shape=(WINDOW_SAMPLES, 1)))
model.add(BatchNormalization())
model.add(MaxPooling1D(pool_size=3))
model.add(Dropout(0.3))

# Hidden Layer 2
model.add(Conv1D(filters=64, kernel_size=5, activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling1D(pool_size=3))
model.add(Dropout(0.3))

# Flatten for Dense layers
model.add(Flatten())

# Dense Layer
model.add(Dense(units=64, activation='relu'))
model.add(Dropout(0.4))

# Output Layer
model.add(Dense(units=N_CLASSES, activation='softmax')) # 3 classes

# Compile the model
# We use 'sparse_categorical_crossentropy' because our labels are integers (0, 1, 2)
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Print the model summary
model.summary()


# --- 5. Model Training and Evaluation ---

print("\nStarting model training...")
history = model.fit(X_train, y_train,
                    epochs=20,
                    batch_size=64,
                    validation_data=(X_test, y_test))

print("\nTraining complete.")

# Evaluate the model on the test set
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"\nTest Accuracy: {accuracy*100:.2f}%")

# Make predictions
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)

# Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_classes, target_names=CLASSES))

# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred_classes)
plt.figure(figsize=(10, 8))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=CLASSES,
            yticklabels=CLASSES)
plt.title('Confusion Matrix - BPM Classification')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()

# Plot training history
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()


# --- 6. Saving the Model ---
model.save('ecg_bpm_model.h5')
print("\nBPM classification model saved as ecg_bpm_model.h5")
```
